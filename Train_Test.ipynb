## Train/Test 
- It is a method to ensure the accuracy of  your model

```python
import numpy as np
import matplotlib.pyplot as plt
import pandas as pd
from sklearn import linear_model
from sklearn.model_selection import train_test_split
from sklearn.metrics import r2_score
```

```python
np.random.seed(2)

x = np.random.normal(3, 1, 100)
y = np.random.normal(150, 40, 100) / x

plt.scatter(x, y)
plt.xlabel("Time spent in shop")
plt.ylabel("Amount spent in dollars")
plt.show()
```

```python
# Split into Train/Test 
train_x = x[:80]
train_y = y[:80]

test_x = x[80:]
test_y = y[80:]
```

```python
#display training set 
plt.scatter(train_x, train_y)
plt.xlabel("Time spent in shop")
plt.ylabel("Amount spent in dollars")
plt.show()
```

```python
#displaying the test set
plt.scatter(test_x,test_y)
plt.xlabel("Time spent in shop")
plt.ylabel("Amount spent in dollars")
plt.show()
```

```python
# a polynomial regression would be the best fit for this data.
my_model = np.poly1d(np.polyfit(train_x, train_y, 4))
my_line = np.linspace(0, 6, 100)
plt.scatter(train_x, train_y)
plt.plot(my_line, my_model(my_line))
plt.xlabel("Time spent in shop")
plt.ylabel("Amount spent in dollars")
plt.show()
```

```python
from sklearn.metrics import r2_score
r2 = r2_score(train_y, my_model(train_x))
print(r2)

#There is a relation between the training data and the model 
```

```python
#We have discovered that the model is good fit for the training data
#now we test the model with the test data
r2_score_test = r2_score(test_y,my_model(test_x))
print(r2_score_test)

print("The model fits the testing set as well therefore we can use the moodel to predict future values")
```

### Project on Train/test 

- I am using the House Sales dataset from kaggle 
- i will use multiple regression to  predict the price of a house based on the number of bedrooms,bathrooms and floors

```python
#loading the dataset 
df = pd.read_csv(r"C:\Users\sanke\Downloads\kc_house_data.csv.zip")
df.head(5)
```


# Performing EDA(Exploratory data analysis)

```python
#converting the date to date time format
df['date'] = pd.to_datetime(df['date'])
df.head(5)
```

```python
#checking for missing values 
df.isnull().sum()
```

```python
#checking for duplicates 
df.duplicated().sum()
```

```python
df.info()
```

- The data is clean 

```python
#descriptive statistics on numerical columns
df.describe()
```

```python
x = df[['bedrooms','bathrooms','floors','sqft_living']]
y = df['price']
```

```python
x_train,x_test,y_train,y_test = train_test_split(x,y,test_size=0.2,random_state=42)  
#random state = 0.2 is basically 20% for testing and 80% for training
```

```python
regression_model = linear_model.LinearRegression()
regression_model.fit(x_train,y_train)
```

```python
y_pred = regression_model.predict(x_test)
```

```python
#checking the accuracy using the r-squared value
r2_score(y_test,y_pred)
```
